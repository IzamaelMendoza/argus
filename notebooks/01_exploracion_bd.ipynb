```python
import os
import glob

carpeta_resfinder = "../data/raw/2_ResFinder/resfinder_db"
archivos_fsa = glob.glob(os.path.join(carpeta_resfinder, "*.fsa"))

print("--- Primeros 5 encabezados de ResFinder ---")
contador = 0
for archivo in archivos_fsa:
    with open(archivo, "r") as f:
        for linea in f:
            if linea.startswith(">"):
                print(f"[{os.path.basename(archivo)}] {linea.strip()}")
                contador += 1
            if contador == 5:
                break
    if contador == 5:
        break
```
